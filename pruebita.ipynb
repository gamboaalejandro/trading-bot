```python
import ccxt
import os

print("--- DIAGNÓSTICO FINAL SPOT TESTNET ---")

# 1. CONFIGURACIÓN
exchange = ccxt.binance({
    'apiKey': os.environ.get('BINANCE_TESTNET_API_KEY', ''),
    'secret': os.environ.get('BINANCE_TESTNET_SECRET', ''),
    'options': {
        'defaultType': 'spot', 

    }
})

exchange.set_sandbox_mode(True)


# 3. EJECUCIÓN
try:
    print("📡 Conectando a Binance Spot Testnet...")
    
    # TRUCO: Pedimos el balance especificando 'type': 'spot'
    # Esto obliga a usar el endpoint /api/v3/account en lugar del que da error 404
    balance = exchange.fetch_balance({'type': 'spot'})
    
    print("\n✅ ¡CONEXIÓN ESTABLECIDA!")
    print("--------------------------------")
    
    has_funds = False
    for asset, amount in balance['total'].items():
        if amount > 0:
            print(f"💰 {asset}: {amount}")
            has_funds = True
            
    if not has_funds:
        print("⚠️ Tu cuenta está vacía (0.00).") 
        print("   Ve a https://testnet.binance.vision/ y usa el Faucet (Login GitHub).")

    symbol = 'BTC/USDT'
    print(f"\n🔎 Buscando precio de {symbol}...")
    ticker = exchange.fetch_ticker(symbol)
    price = ticker['last']
    print(f"   Precio actual: ${price:,.2f}")
    # Orden fija de $50 USD
    order_value_usdt = 50
    buy_price = price * 0.99  # 99% del precio actual
    btc_quantity = order_value_usdt / buy_price
    
    print(f"\n🚀 Enviando orden LIMIT de PRUEBA:")
    print(f"   Cantidad: {btc_quantity:.6f} BTC")
    print(f"   Precio: ${buy_price:,.2f}")
    print(f"   Valor total: ${order_value_usdt} USDT")
    
    order = exchange.create_order(symbol, 'limit', 'buy', btc_quantity, buy_price)
    
    print(f"\n✅ ¡Orden creada exitosamente!")
    print(f"   ID: {order['id']}")
    print(f"   Status: {order['status']}")
    print(f"✅ Orden Exitosa! ID: {order['id']}")

except ccxt.NetworkError as e:
    print(f"\n❌ Error de Red: {e}")
except ccxt.ExchangeError as e:
    print(f"\n❌ Error del Exchange (Lógica): {e}")
except Exception as e:
    print(f"\n❌ Error General: {e}")
```
